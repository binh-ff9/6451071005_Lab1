

# Đọc dữ liệu và one-hot encoding

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

# Mỗi dòng là một giao dịch, các mặt hàng cách nhau bởi dấu phẩy
transactions = []
with open("groceries.csv", encoding="utf-8") as f:
    for line in f:
        items = [x.strip() for x in line.strip().split(",") if x.strip()]
        if items:
            transactions.append(items)

print("Số giao dịch:", len(transactions))   # kỳ vọng: 9835

te = TransactionEncoder()
onehot = te.fit(transactions).transform(transactions)
df = pd.DataFrame(onehot, columns=te.columns_).astype(bool)
print("Số items:", df.shape[1])             # kỳ vọng: 169
df.head()
```

# Chạy Apriori (a) và FP-Growth (b), đo thời gian

```python
import time
import gc
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, fpgrowth

min_sup_values = [i / 10000 for i in range(1, 11)]

apriori_results = []
fpgrowth_results = []

for min_sup in min_sup_values:
    print(f"\n===== min_sup = {min_sup * 100:.2f}% =====")

    # APRIORI
    start = time.perf_counter()

    result_a = apriori(
        df,
        min_support=min_sup,
        use_colnames=True,
        low_memory=True
    )

    time_a = time.perf_counter() - start

    apriori_results.append({
        "min_sup (%)": min_sup * 100,
        "Số tập phổ biến": len(result_a),
        "Thời gian (giây)": time_a
    })

    result_a["itemsets"] = result_a["itemsets"].apply(
        lambda s: ", ".join(sorted(s))
    )
    result_a.to_csv(
        f"/content/apriori_{min_sup:.4f}.csv",
        index=False
    )

    print(f"Apriori: {len(result_a)} tập, {time_a:.4f} giây")

    del result_a
    gc.collect()

    # FP-GROWTH
    start = time.perf_counter()

    result_f = fpgrowth(
        df,
        min_support=min_sup,
        use_colnames=True
    )

    time_f = time.perf_counter() - start

    fpgrowth_results.append({
        "min_sup (%)": min_sup * 100,
        "Số tập phổ biến": len(result_f),
        "Thời gian (giây)": time_f
    })

    result_f["itemsets"] = result_f["itemsets"].apply(
        lambda s: ", ".join(sorted(s))
    )
    result_f.to_csv(
        f"/content/fpgrowth_{min_sup:.4f}.csv",
        index=False
    )

    print(f"FP-Growth: {len(result_f)} tập, {time_f:.4f} giây")

    del result_f
    gc.collect()

# Bảng tổng hợp
apriori_df = pd.DataFrame(apriori_results)
fpgrowth_df = pd.DataFrame(fpgrowth_results)

display(apriori_df)
display(fpgrowth_df)

apriori_df.to_csv("/content/apriori_timing.csv", index=False)
fpgrowth_df.to_csv("/content/fpgrowth_timing.csv", index=False)
```

# Vẽ biểu đồ so sánh (c)

```python
comparison = pd.DataFrame({
    "min_sup (%)": apriori_df["min_sup (%)"],
    "Apriori": apriori_df["Thời gian (giây)"],
    "FP-Growth": fpgrowth_df["Thời gian (giây)"]
})

plt.figure(figsize=(10, 6))
plt.plot(
    comparison["min_sup (%)"],
    comparison["Apriori"],
    marker="o",
    label="Apriori"
)
plt.plot(
    comparison["min_sup (%)"],
    comparison["FP-Growth"],
    marker="s",
    label="FP-Growth"
)

plt.xlabel("Minimum Support (%)")
plt.ylabel("Execution Time (seconds)")
plt.title("Apriori vs FP-Growth on Groceries Dataset2")
plt.xticks(comparison["min_sup (%)"])
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig("/content/algorithm_comparison.png", dpi=300)
plt.show()

comparison.to_csv("/content/comparison_results.csv", index=False)
```
